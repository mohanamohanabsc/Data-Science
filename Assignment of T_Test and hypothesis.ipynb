```python
import pandas as pd
```

```python
dataset=pd.read_csv("Placementdata.csv")
dataset
```

```python
dataset.isna().sum()
```

```python
#Assignment: Analzis for Commerce and Science student in mba_p
#Unpaird T-Test(different group with same condition)
```

```python
from scipy.stats import ttest_ind
dataset=dataset.dropna()
com=dataset[dataset["hsc_s"]=='Commerce']["mba_p"]
sci=dataset[dataset["hsc_s"]=='Science']["mba_p"]
ttest_ind(com,sci)
```

```python
#Assignment: Analzis for Commerce student in mba_p and etest_p 
#Paird T-Test(same group with different condition)
```

```python
from scipy.stats import ttest_rel
dataset=dataset.dropna()
com=dataset[dataset["hsc_s"]=='Commerce']["mba_p"]
comm=dataset[dataset["hsc_s"]=='Commerce']["etest_p"]
ttest_rel(com,comm)
```

```python
#Reject Null hypothesis p<0.05
#Accept Null hypothesis and Reject alternate hypothesis
#There is no similarity between mba_p and etest_p of commerce student
```
